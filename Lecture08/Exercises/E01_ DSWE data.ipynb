### Dataset description

From Data Science for Wind Energy, by Yu Ding, chapter 2: https://aml.engr.tamu.edu/book-dswe/

Data: https://zenodo.org/records/5516539

The data corresponds to the wind speed and power produced by a single wind turbine within an onshore wind farm. The data covers a full year, at hourly or 10-minute temporal resolution. We focus here on predicting future wind speed data, using time series forecasting methods.

Key characteristics of the data are its daily (diurnal) and annual seasonality; wind speeds are typically greater during the daytime owing to air currents driven by heating, and are also typically greater during winter. It is important to account for both of these effects in developing a time series forecasting model.

Forecasting the wind speed is essential to the integration of wind power into the energy grid. Unlike conventional (e.g. fossil fuel) generators, wind power is highly variable, and we have almost no control over the amount of wind power generated (other than to turn wind turbines down or off completely - this wastes "free" energy). Conventionally, numerical weather prediction (NWP) models, or more recently large spatio-temporal machine learning-based models, are used to produce wind forecasts at time scales of hours or days ahead. In this practical we explore the applicability of pure time-series forecasting methods, and specifically the ARMA model.
