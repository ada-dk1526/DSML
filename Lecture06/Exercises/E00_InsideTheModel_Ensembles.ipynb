### Lecture connection and short recap

Revisit **Lecture 6 → “Ensemble Methods” → “Parallel Methods”**. The lecture describes combining classification decisions by voting. Its later `VotingClassifier` example uses **soft voting**, which averages probabilities. Here we use **hard voting**, using only the predicted classes.

This exercise studies three equal hard votes from already fitted classifiers. Bagging, random forests, boosting and stacking differ in how models are trained and combined; this table isolates the hard-vote step.

Each of three classifiers casts one vote for class 0 or class 1. The class with at least two votes wins. All votes have equal weight; three binary votes cannot tie. We inspect predictions already made on the same six labelled cases; we do not train, tune or select a model on them.

**Reasoning aid:** a classifier's accuracy says how often it is right. A table of predictions also shows *which cases* it gets right. Keep these two kinds of information separate.

## 4. Verify with two short calculations

**About 8 minutes · Python now**

Complete the two marked expressions. `votes` should contain the number of votes for class 1 in each case; `ensemble` should contain the resulting 0/1 labels. The same two lines run for both versions of C. NumPy adds these arrays element by element; `.astype(int)` can convert Boolean decisions to 0/1.

The supplied plot counts **wrong votes on each case**, not votes for class 1. Check that you can explain the difference between those quantities.


```python
import numpy as np
import matplotlib.pyplot as plt

y = np.array([0, 0, 0, 1, 1, 1])
pred_A = np.array([1, 1, 0, 1, 1, 1])
pred_B = np.array([0, 0, 1, 0, 1, 1])
pred_C = np.array([0, 0, 0, 1, 0, 0])
pred_C_new = np.array([1, 1, 0, 1, 1, 1])
```

```python
ensemble_predictions = []
wrong_vote_counts = []
names = ['Original C', 'Changed C']

for name, c in zip(names, [pred_C, pred_C_new]):
    votes = None  # ADD YOUR CODE HERE: add A, B and this version of C.
    ensemble = None  # ADD YOUR CODE HERE: convert the majority decision to 0/1.
    if votes is None or ensemble is None:
        print('Complete the two expressions, then run this cell again.')
        break

    wrong_A = (pred_A != y).astype(int)
    wrong_B = (pred_B != y).astype(int)
    wrong_C = (c != y).astype(int)
    wrong_count = wrong_A + wrong_B + wrong_C
    ensemble_predictions.append(ensemble)
    wrong_vote_counts.append(wrong_count)
    print(name)
    print('Correct out of 6: A, B, C =', np.sum(pred_A == y),
          np.sum(pred_B == y), np.sum(c == y))
    print('Votes for class 1:', votes)
    print('Ensemble labels: ', ensemble)
    print('Ensemble correct:', np.sum(ensemble == y), 'out of 6')
    print('Wrong votes per case:', wrong_count)
    print()
```

```python
if len(ensemble_predictions) == 2:
    fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharey=True)
    cases = np.arange(1, 7)
    for ax, name, counts in zip(axes, names, wrong_vote_counts):
        ax.bar(cases, counts, color='tab:blue', width=0.65)
        ax.axhline(2, color='red', linestyle='--', label='Wrong majority: 2 or more')
        for case, count in zip(cases, counts):
            ax.text(case, count + 0.08, str(count), ha='center')
        ax.set_title(name)
        ax.set_xlabel('Case')
        ax.set_xticks(cases)
        ax.set_yticks([0, 1, 2, 3])
        ax.set_ylim(0, 3.3)
        ax.grid(axis='y', alpha=0.2)
        ax.legend(loc='upper right', fontsize=8)
    axes[0].set_ylabel('Number of wrong votes')
    plt.tight_layout()
    plt.show()
```
