#### Interpreting Your Curves

<details>
<summary>💡 Make your own interpretation before opening this</summary>

- With a small subset, training error may be lower than validation error. The curve starts at **10 observations**, so a zero training error is not guaranteed.
- A large validation–training gap suggests sensitivity to the fitted sample; it does not by itself identify the cause.
- As sample size increases, training error often rises and validation error often falls, but neither trend must be monotonic for one sequence of subsets.
- Similar, high errors can suggest limited model flexibility, missing information or substantial response noise. Similar errors alone do not prove that a model is fit for purpose.
- Flattening sample-size curves concerns the benefit of additional data. It is not evidence that an iterative optimiser has converged. Judge acceptable error using porosity units and the intended application.
