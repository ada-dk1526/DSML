## Regularization 🎯

Use the supplied house-sales dataset to compare Linear Regression, Ridge, Lasso and Elastic Net. The supplied preprocessing scaffold lets you focus on model comparison; categorical encoding will be explained in more detail in Lecture 3.

The file contains accidental duplicate records and an extra `Pesos` column equal to 20 times `SalePrice` wherever it is observed. That column directly reveals the response, so it must not be a predictor. The scaffold removes it, the row identifier and exact duplicate records before splitting.

Use training data to fit, validation data to compare models, and the test set once for the final selected model.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline
from sklearn.base import clone

houses_data = pd.read_csv('../../Excercise_data/L02/ML_Houses_dataset.csv')
houses_data = houses_data.drop_duplicates().dropna(subset=['SalePrice'])
X_houses = houses_data.drop(columns=['SalePrice', 'Id', 'Pesos'])
y_houses = houses_data['SalePrice']

# Reserve test data, then create a validation set from development data.
X_dev_houses, X_test_houses, y_dev_houses, y_test_houses = train_test_split(
    X_houses, y_houses, test_size=0.2, random_state=42)
X_train_houses, X_val_houses, y_train_houses, y_val_houses = train_test_split(
    X_dev_houses, y_dev_houses, test_size=0.2, random_state=42)

num_cols = X_train_houses.select_dtypes(include='number').columns
cat_cols = X_train_houses.select_dtypes(exclude='number').columns
numeric = make_pipeline(
    SimpleImputer(strategy='median'), StandardScaler())
categorical = make_pipeline(
    SimpleImputer(strategy='constant', fill_value='missing'),
    OneHotEncoder(handle_unknown='ignore'))
preprocessor = ColumnTransformer([
    ('numeric', numeric, num_cols),
    ('categorical', categorical, cat_cols)
])

# Unfitted instructions: fit them only inside each model's pipeline.
X_train_houses.shape, X_val_houses.shape, X_test_houses.shape
```
