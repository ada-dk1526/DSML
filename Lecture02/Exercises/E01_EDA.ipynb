# 🔄 Scale Numerical Features

Choose a scaler using the model's behaviour, feature units and sensitivity to extreme values:

- `StandardScaler` centres and scales features using training means and standard deviations. It does not require normal predictors or make skewed data normal.
- `MinMaxScaler` maps the observed training range to [0, 1]. It preserves skewness, is sensitive to extremes, and can map new values outside [0, 1].
- `RobustScaler` uses training medians and interquartile ranges. Its scale estimates are less sensitive to extremes; it does not remove outliers or make the model itself robust.
