# Machine Learning & Data Science Road Map

## 1. Data Preprocessing  <span style="font-size: 2em;">🧹</span>

- **Data Cleaning**
  - Handling duplicates
  - Splitting the data into training, validation, and test sets
  - Handling missing data
  - Identifying and dealing with outliers
  - Data scaling and normalization
- **Exploratory Data Analysis (EDA)**
  - Data visualization
  - Statistical summaries
  - Identifying patterns and outliers
- **Feature Engineering**
  - Creating new features
  - Interaction terms and domain-specific features
  - Encoding categorical variables
  - Collinearity handling
- **Dimensionality Reduction**
  - PCA
  - t-SNE
  - UMAP
- **Data Imbalance**
  - Techniques: undersampling, oversampling
- **Pipelines**
  - Structuring ML projects
  - Automating preprocessing steps
  - Ensuring consistency in workflows
- **timeseries**
  - Datetime Resampling, Shifting, Rolling, Expanding
  - Stationarity, Trend, Seasonality, Time aware feature engineering
  - ARMA, ARIMA, SARIMA, Decision TREE

## 2. Machine Learning Algorithms  <span style="font-size: 2em;">📈</span>

- **Supervised Learning**
  - **Regression**
    - Linear Regression
    - Ridge/Lasso Regression
  - **Classification**
    - Logistic Regression
    - K-Nearest Neighbors
    - Support Vector Machines
    - Decision Trees and Random Forests
    - Gradient Boosting Machines (GBM, XGBoost)

- **Unsupervised Learning**
  - **Clustering**
    - K-Means
    - DBSCAN
  - **Dimensionality Reduction**
    - PCA
    - t-SNE
    - UMAP
  - **Anomaly Detection**

- **Ensemble Learning**
  - Bagging
  - Boosting
  - Stacking
  - Voting Classifiers

## 3. Model Selection and Evaluation <span style="font-size: 2em;">🕵️</span>

- **Performance Metrics**
  - Classification Metrics (Accuracy, Precision, Recall, F1-Score)
  - Regression Metrics (MSE, RMSE, MAE)
  - ROC and AUC
  - Precision-Recall Curve
  - Bias-Variance Tradeoff
  - Overfitting vs. Underfitting
  - Model Complexity

- **Model Validation**
  - Cross-Validation (K-Fold, Stratified K-Fold)
- **Hyperparameter Tuning**
  - Grid Search
  - Random Search

- **Model Interpretability**
  - Feature Importance
  - SHAP Values
  - LIME

## TimeSeries:
- Introduction to Time Series
- Trend and Stationarity
- Seasonality
- Modeling the Stationary Component
- ACF/PACF
- From ARMA → ARIMA → SARIMA → SARIMAX
- Evaluation and Model Selection (AIC, BIC)
