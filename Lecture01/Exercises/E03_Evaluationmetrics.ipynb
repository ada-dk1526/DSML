## Interpretation of Each Metric

### Mean Absolute Error (MAE)
This metric gives the average magnitude of the errors in the model's predictions, without considering their direction. It tells us the average error in the same units as the target variable. you should think about what an acceptable average error might be in the context of the problem.

### Mean Squared Error (MSE)
Unlike MAE, MSE penalizes larger errors more heavily due to squaring. you will understand that it amplifies the impact of large errors, which can be useful when such errors are especially costly.

### Root Mean Squared Error (RMSE)
RMSE is the square root of MSE and provides an error metric in the same units as the target variable. It's often easier to interpret than MSE because it’s more intuitive. However, like MSE, it penalizes larger errors more.

### R² (Coefficient of Determination)
This metric explains how well the regression model fits the data. A score of 1 means the model perfectly explains all the variability of the target data, while a score closer to 0 indicates that the model does not explain much of the variance. you should interpret R² relative to how complex the dataset is and how much variance they expect to explain.

## Max Error

Maximum absolute error gives the size of the largest error, but not its direction.

Find the observation with the largest $|y_i-\hat y_i|$ and compare its measured and predicted `rspm` values. Define the signed residual as $r_i=y_i-\hat y_i$:

- $r_i>0$: the model underpredicted this observation.
- $r_i<0$: the model overpredicted this observation.

Is the error unusually large in the context of the measured concentration? A large residual is a reason to investigate; it does not by itself establish that the observation is an outlier or reveal why the model failed.
