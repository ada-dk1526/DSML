<details>
  <summary>💡 Hint</summary>

1. Remember from the class that we will use the `fit()` function on our imputer class to learn the statistics from our `X_train`, and then use the `transform()` function to change our data! So:
> - `fit_transform` on **Training Data**: The `fit_transform` method calculates the required statistic (the impute strategy) and applies the transformation to the training data.
> - `transform` on **Test Data**: The transform method applies the same transformation (using the statistic calculated from the training data) to the test data. This ensures consistency and prevents data leakage.** 

  
 </details>
