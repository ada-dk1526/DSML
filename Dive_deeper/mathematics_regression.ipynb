# Linear Regression and the Normal Equation

Let’s say we have a matrix $ \mathbf{X} $ of features and a target vector $ \mathbf{y} $. The goal of linear regression is to find the coefficients $ \boldsymbol{\theta} $ that best relate $ \mathbf{X} $ to $ \mathbf{y} $. A direct way to solve for $ \boldsymbol{\theta} $ would be:

$$ \boldsymbol{\theta} = \mathbf{X}^{-1} \mathbf{y} $$

However, this isn’t generally feasible for two main reasons:

1. Most of the time, our $ \mathbf{X} $ matrix is not square (it has more rows than columns), meaning $ \mathbf{X} $ is not invertible.
2. In most cases, we have more data points than features (an overdetermined system), so there’s no exact solution for $ \mathbf{X} \boldsymbol{\theta} = \mathbf{y} $.

Instead, we seek the best approximate solution by minimizing the Mean Squared Error (MSE). For linear regression, the MSE represents the average squared difference between the actual values $ \mathbf{y} $ and the predicted values $ \hat{\mathbf{y}} = \mathbf{X} \boldsymbol{\theta} $. Mathematically, we write:

$$ \text{MSE} = \frac{1}{2m} \sum_{i=1}^{m} (y_i - \hat{y}_i)^2 = \frac{1}{2m} (\mathbf{y} - \mathbf{X} \boldsymbol{\theta})^T (\mathbf{y} - \mathbf{X} \boldsymbol{\theta}) $$

where:
- $ m $ is the number of data points,
- $ y_i $ are the actual target values, and
- $ \hat{y}_i = \mathbf{X} \boldsymbol{\theta} $ are the predicted values.

Our goal is to find the $ \boldsymbol{\theta} $ that minimizes this MSE.

## Derivation of the Normal Equation

To find $ \boldsymbol{\theta} $ that minimizes the MSE, we take the derivative of the cost function with respect to $ \boldsymbol{\theta} $, set it to zero, and solve. Expanding and simplifying the expression for the MSE leads us to the normal equation:

$$ \boldsymbol{\theta} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y} $$

where:
- $ \mathbf{X}^T \mathbf{X} $: captures relationships between features in $ \mathbf{X} $ (like feature correlations).
- $ \mathbf{X}^T \mathbf{y} $: captures the relationship between each feature and the target variable $ \mathbf{y} $.
- $ (\mathbf{X}^T \mathbf{X})^{-1} $: inverts the relationship matrix, allowing us to solve for $ \boldsymbol{\theta} $.

This normal equation finds the values of $ \boldsymbol{\theta} $ that best fit the data, minimizing the MSE.

## Extensions to Address Multicollinearity

Once we have this least-squares solution, we can extend it by adding a penalty term to the cost function to deal with multicollinearity (when features are highly correlated). Regularized regression methods like ridge and lasso regression add such penalties, leading to more stable solutions and better generalization to new data.

In summary, the normal equation provides a powerful way to find the least-squares solution for linear regression by minimizing the MSE, even when a direct inversion of $ \mathbf{X} $ isn’t possible.


## Why Use $\frac{1}{2m}$ Instead of $\frac{1}{m}$?

#### $\frac{1}{m}$:
- This is the straightforward average of squared errors, representing the Mean Squared Error.

#### $\frac{1}{2m}$:
- This form is often used in optimization problems because it simplifies the derivative by removing an extra factor of 2, making the gradient calculation cleaner.

### Derivative Simplification
When we set up the MSE as:

$$
\text{MSE} = \frac{1}{2m} \sum_{i=1}^{m} (y_i - \hat{y}_i)^2
$$

then the derivative with respect to $\theta$ doesn’t include an extra factor of 2:

$$
\frac{d}{d\theta} \left( \frac{1}{2m} \sum_{i=1}^{m} (y_i - \hat{y}_i)^2 \right) = \frac{1}{m} \sum_{i=1}^{m} (y_i - \hat{y}_i) \cdot (-X_i)
$$

This simplification is particularly useful in gradient descent and other optimization algorithms, as it makes the math cleaner and the implementation more straightforward.


# Ridge Regression: Step-by-Step Solution

Let’s walk through the mathematical solution of ridge regression using a small data set, focusing on how we compute the coefficients step by step.

## Step 1: Set Up a Small Dataset
Let's assume we have a dataset with two features ($X_1$ and $X_2$) and an outcome variable $y$:

| $X_1$ | $X_2$ | $y$ |
|------|------|----|
| 1    | 2    | 5  |
| 2    | 3    | 7  |
| 3    | 4    | 9  |

Here:
- $X$ is our matrix of features.
- $y$ is the vector of target values.

In matrix form, we write:

$$
X = \begin{bmatrix}
1 & 2 \\
2 & 3 \\
3 & 4
\end{bmatrix},
\quad
y = \begin{bmatrix}
5 \\
7 \\
9
\end{bmatrix}
$$

## Step 2: Ridge Regression Cost Function
Ridge regression minimizes a modified cost function:

$$
\text{Cost Function} = \text{MSE} + \alpha \sum_{j=1}^{p} \theta_j^2
$$

This regularization term, $\alpha \sum_{j=1}^{p} \theta_j^2$, penalizes large coefficients, where:
- $\alpha$ is the regularization strength.
- $\theta_j$ are the coefficients of the features $X_j$.

## Step 3: Normal Equation for Ridge Regression
To find the optimal coefficients $\theta$, ridge regression solves:

$$
\theta = (X^T X + \alpha I)^{-1} X^T y
$$

where:
- $X^T$ is the transpose of $X$.
- $I$ is the identity matrix, ensuring the regularization term is applied.
- $\alpha$ controls the degree of regularization.

Let's break down each part.

## Step 4: Compute $X^T X$
Calculate the matrix product $X^T X$:

$$
X^T = \begin{bmatrix}
1 & 2 & 3 \\
2 & 3 & 4
\end{bmatrix}
$$

Then,

$$
X^T X = \begin{bmatrix}
1 & 2 & 3 \\
2 & 3 & 4
\end{bmatrix}
\begin{bmatrix}
1 & 2 \\
2 & 3 \\
3 & 4
\end{bmatrix}
= \begin{bmatrix}
14 & 20 \\
20 & 29
\end{bmatrix}
$$

## Step 5: Add Regularization Term $\alpha I$
Assume $\alpha = 1$ (you can adjust this as desired). The identity matrix $I$ for a 2x2 matrix is:

$$
I = \begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
$$

Then,

$$
\alpha I = \begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
$$

Add $\alpha I$ to $X^T X$:

$$
X^T X + \alpha I = \begin{bmatrix}
14 & 20 \\
20 & 29
\end{bmatrix}
+ \begin{bmatrix}
1 & 0 \\
0 & 1
\end{bmatrix}
= \begin{bmatrix}
15 & 20 \\
20 & 30
\end{bmatrix}
$$

## Step 6: Compute $(X^T X + \alpha I)^{-1}$
Now, we need to invert $\begin{bmatrix} 15 & 20 \\ 20 & 30 \end{bmatrix}$. For simplicity, let's find the inverse by hand:

The inverse of a 2x2 matrix $\begin{bmatrix} a & b \\ c & d \end{bmatrix}$ is given by:

$$
\frac{1}{ad - bc} \begin{bmatrix} d & -b \\ -c & a \end{bmatrix}
$$

So here:

$$
\text{det} = (15)(30) - (20)(20) = 450 - 400 = 50
$$

The inverse is:

$$
(X^T X + \alpha I)^{-1} = \frac{1}{50} \begin{bmatrix} 30 & -20 \\ -20 & 15 \end{bmatrix} = \begin{bmatrix} 0.6 & -0.4 \\ -0.4 & 0.3 \end{bmatrix}
$$

## Step 7: Compute $X^T y$
Now, calculate $X^T y$:

$$
X^T y = \begin{bmatrix}
1 & 2 & 3 \\
2 & 3 & 4
\end{bmatrix}
\begin{bmatrix}
5 \\
7 \\
9
\end{bmatrix}
= \begin{bmatrix}
46 \\
68
\end{bmatrix}
$$

## Step 8: Compute $\theta = (X^T X + \alpha I)^{-1} X^T y$
Multiply $(X^T X + \alpha I)^{-1}$ by $X^T y$:

$$
\theta = \begin{bmatrix}
0.6 & -0.4 \\
-0.4 & 0.3
\end{bmatrix}
\begin{bmatrix}
46 \\
68
\end{bmatrix}
$$

Breaking down the multiplication:

First row: $(0.6)(46) + (-0.4)(68) = 27.6 - 27.2 = 0.4$

Second row: $(-0.4)(46) + (0.3)(68) = -18.4 + 20.4 = 2.0$

So, $\theta \approx \begin{bmatrix} 0.4 \\ 2.0 \end{bmatrix}$.

## Summary of Results
The optimal coefficients $\theta$ for this ridge regression example are:

$$
\theta_0 = 0.4, \quad \theta_1 = 2.0
$$

These values minimize the ridge regression cost function for the given data set and regularization parameter $\alpha = 1$.
