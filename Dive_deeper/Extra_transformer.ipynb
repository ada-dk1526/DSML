```python
import pandas as pd
from sklearn.impute import SimpleImputer

X_train = pd.DataFrame({'Feature1': [1.0, 2.0, None, 4.0], 'Feature2': [2.0, None, 4.0, 5.0]})
X_test = pd.DataFrame({'Feature1': [5.0, None], 'Feature2': [6.0, 7.0]})
```

```python
X_train
```

```python
X_test
```

```python
imputer = SimpleImputer(strategy='mean')

# Fit and transform X_train
X_train_imputed = imputer.fit_transform(X_train)

# Transform X_test using the same imputer
X_test_imputed = imputer.transform(X_test)
```

```python
X_train_imputed_df = pd.DataFrame(X_train_imputed, columns=['Feature1', 'Feature2'])
X_test_imputed_df = pd.DataFrame(X_test_imputed, columns=['Feature1', 'Feature2'])

print("Imputed X_train:")
print(X_train_imputed_df)
print("\nImputed X_test:")
print(X_test_imputed_df)
```

- The `fit_transform()` method on `X_train` computes the `mean` of the columns and fills `NaN` values with these computed means.

- The `.transform()` method on `X_test` applies the imputation based on the statistics learned from `X_train` without recalculating the mean from `X_test` and only fills `NaN` values with `X_train` computed means.

- Why we are not calculating the mean of the `X_test` then?

  Avoid Data Leakage: Using `fit_transform()` on `X_test` would allow the model to "see" information from the test set, leading to overfitting and an unrealistic evaluation of the model's performance and we want to avoid that.
