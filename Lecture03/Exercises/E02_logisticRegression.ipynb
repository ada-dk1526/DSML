## Why Use StratifiedKFold for Classification?
Stratification keeps class proportions approximately similar across folds. It helps avoid folds with too few minority-class observations, but does not remove class imbalance or dependence between observations.

Use training-only cross-validation if comparing models. Where multiple rows belong to the same patient, group-aware splitting takes priority over ordinary row stratification.
