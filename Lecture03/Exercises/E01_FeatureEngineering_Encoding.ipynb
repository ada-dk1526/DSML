Regularization Effects:

- Linear Regression serves as a baseline with no regularization.
- Ridge penalizes large coefficients, which helps to reduce overfitting. It can help when predictors are strongly correlated, but improved held-out performance is not guaranteed.
- Lasso can shrink some coefficients to zero, effectively performing feature selection. It’s useful when you suspect many features have little or no impact on the target variable.
- ElasticNet combines Lasso and Ridge, balancing between reducing large coefficients and setting some coefficients to zero. It’s useful when you want the benefits of both L1 and L2 regularization.
</details>
